```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats

df = pd.read_csv("c:\My project857\Shopping Trends And Customer Behaviour Dataset.csv")


plt.style.use('default')


def plot_least_squares_trend(
    x_col, y_col, x_label, y_label, title, num_bins=10
):
 
  bins = pd.cut(df[x_col], bins=num_bins)
  grouped = df.groupby(bins, observed=False)[y_col].mean().reset_index()

  
  x_mids = np.array([interval.mid for interval in grouped[x_col]])
  y_vals = grouped[y_col].values
  x_labels = [
      f'{int(interval.left)}-{int(interval.right)}' for interval in grouped[x_col]
  ]

  
  slope, intercept, r_value, p_value, std_err = stats.linregress(x_mids, y_vals)

  x_numeric = np.arange(len(x_mids))
  slope_idx, intercept_idx, _, _, _ = stats.linregress(x_numeric, y_vals)
  line_y = slope_idx * x_numeric + intercept_idx

  
  fig, ax = plt.subplots(figsize=(10, 6), facecolor='#f5f5ed')
  ax.set_facecolor('#f5f5ed')

  
  ax.scatter(
      x_numeric, y_vals, color='blue', s=45, zorder=3, label='Average Data'
  )

  
  sign = '+' if intercept >= 0 else '-'
  label_text = (
      f'Least Squares Line (y = {slope:.2f}x {sign} {abs(intercept):.2f})'
  )
  ax.plot(
      x_numeric, line_y, color='red', linewidth=2, zorder=2, label=label_text
  )

 
  ax.set_title(title, fontsize=14, fontweight='bold', pad=15)
  ax.set_xlabel(x_label, fontsize=12)
  ax.set_ylabel(y_label, fontsize=12)

  
  ax.set_xticks(x_numeric)
  ax.set_xticklabels(x_labels, rotation=30, ha='right', fontsize=9)

  
  ax.grid(axis='y', linestyle='--', alpha=0.5, color='gray')

  
  ax.legend(
      loc='upper left',
      frameon=True,
      facecolor='white',
      edgecolor='gray',
      fontsize=10,
  )

  plt.tight_layout()
  plt.show()




# ชุดที่ 1: Age vs Purchase Amount
plot_least_squares_trend(
    'Age',
    'Purchase Amount (USD)',
    'Age (Years)',
    'Average Purchase Amount (USD)',
    'Linear Trend of Purchase Amount by Age (Least Squares)',
    num_bins=10,
)

# ชุดที่ 2: Previous Purchases vs Review Rating
plot_least_squares_trend(
    'Previous Purchases',
    'Review Rating',
    'Previous Purchases (Times)',
    'Average Review Rating',
    'Linear Trend of Review Rating by Previous Purchases (Least Squares)',
    num_bins=10,
)

# ชุดที่ 3: Purchase Amount vs Review Rating
plot_least_squares_trend(
    'Purchase Amount (USD)',
    'Review Rating',
    'Purchase Amount (USD)',
    'Average Review Rating',
    'Linear Trend of Review Rating by Purchase Amount (Least Squares)',
    num_bins=10,
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("c:\My project857\Shopping Trends And Customer Behaviour Dataset.csv")


sales = (df.groupby("Category")["Purchase Amount (USD)"]
           .sum()
           .sort_values(ascending=False))

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(sales.index, sales.values, color="#2a78d6", width=0.6)
ax.bar_label(bars, labels=[f"${v:,.0f}" for v in sales.values], padding=3)

ax.set_title("Total sales by category", fontweight="bold", loc="left")
ax.set_ylabel("Total purchase amount (USD)")
ax.spines[["top", "right"]].set_visible(False)
ax.yaxis.grid(True, color="#e5e4e0")
ax.set_axisbelow(True)

plt.tight_layout()
plt.savefig("category_sales.png", dpi=150)
plt.show()
```
